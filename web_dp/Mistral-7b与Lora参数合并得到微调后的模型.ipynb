```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel
```

```python
# 路径：替换为你自己的
BASE_MODEL = "../../model/Mistral-7B/mistralai/Mistral-7B-Instruct-v0___3/"          # 原始模型文件夹
LORA_PATH = "../../output/VLE_LLM/等温等压/体系划分/random_82_aug_fine/checkpoint-1140"      # 你的LoRA权重目录
MERGED_PATH = "merged_model"       # 合并后保存路径
```

```python
# 1. 加载基座模型
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    dtype="auto",
    device_map="auto"
)
```

```python
# 2. 加载LoRA权重
model = PeftModel.from_pretrained(base_model, LORA_PATH)
```

```python
# 3. 合并权重
model = model.merge_and_unload()   # 把LoRA权重合并回原模型
```

```python
# 4. 保存合并后的模型
model.save_pretrained(MERGED_PATH)
# 保存分词器（关键补充）
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)
tokenizer.save_pretrained(MERGED_PATH)
print(f"✅ 已成功合并并保存至: {MERGED_PATH}")
```
